```python
import sys
import runpy

# 2. eval.py에 전달할 인자(Arguments)를 설정합니다.
# 주피터 노트북 파일이 eval.py와 같은 경로에 있으므로 기준 경로는 './'가 됩니다.
sys.argv = [
    "eval.py",
    "--model", "./weights/model-best.pth",       # 가중치 파일 경로
    "--infos_path", "./weights/infos_fc-best.pkl",   # 정보 파일 경로
    "--image_folder", "./inputs",               # 인풋 이미지가 들어있는 폴더 경로
    "--num_images", "-1"                         # 추론할 이미지 개수 (-1로 설정하면 폴더 내 전체 이미지 처리)
]

# 3. eval.py 스크립트를 현재 주피터 커널 환경에서 실행합니다.
print("이미지 캡셔닝 추론을 시작합니다...")
try:
    # eval.py의 코드를 __main__ 환경으로 실행시킵니다.
    runtime_env = runpy.run_path("eval.py", run_name="__main__")
    print("추론이 완료되었습니다. 결과는 'vis/vis.json' 파일로 저장됩니다.")
except Exception as e:
    print(f"실행 중 오류가 발생했습니다: {e}")
```
