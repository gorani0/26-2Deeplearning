```python
weights = torch.tensor([0.2126, 0.7152, 0.0722])
```

```python
unsqueezed_weights = weights.unsqueeze(-1).unsqueeze_(-1)
unsqueezed_weights.shape
```

```python
batch_weights = (batch_t * unsqueezed_weights) #[2, 3, 5, 5] * [3, 1, 1]
batch_gray_weights = batch_weights.mean(-3)
batch_weights.shape, batch_gray_weights.shape
```

```python
import time

# 1. 대규모 연산을 위한 대형 행렬 크기 설정 (10000 x 10000)
matrix_size = 10000

# CPU에서 무작위 텐서 생성
x_cpu = torch.randn(matrix_size, matrix_size)
y_cpu = torch.randn(matrix_size, matrix_size)

# ----------------------------------------
# [1] CPU 행렬 곱셈 속도 측정
# ----------------------------------------
start_cpu = time.time()
result_cpu = torch.matmul(x_cpu, y_cpu)
end_cpu = time.time() - start_cpu
print(f" CPU 행렬 곱셈 소요 시간: {end_cpu:.4f} 초")

```

```python
# ----------------------------------------
# [2] GPU 행렬 곱셈 속도 측정
# ----------------------------------------
# 텐서를 GPU 영역('cuda')으로 명시적 이동
x_gpu = x_cpu.to('cuda')
y_gpu = y_cpu.to('cuda')

# GPU Warm-up (첫 연산은 CUDA 컨텍스트 초기화 오버헤드가 있으므로 측정에서 제외)
_ = torch.matmul(x_gpu, y_gpu)
torch.cuda.synchronize()  # 앞선 초기화 연산이 완전히 끝날 때까지 대기

# 실제 GPU 속도 측정 시작
start_gpu = time.time()
result_gpu = torch.matmul(x_gpu, y_gpu)

#  GPU 연산은 비동기(Asynchronous)로 처리되므로, 
# 연산이 최종 완료될 때까지 CPU 타이머를 대기
torch.cuda.synchronize() 
end_gpu = time.time() - start_gpu
print(f" GPU 행렬 곱셈 소요 시간: {end_gpu:.4f} 초")

# ----------------------------------------
# [3] 결과 비교 출력
# ----------------------------------------
speedup = end_cpu / end_gpu
print(f"\n 대규모 연산에서 GPU가 CPU보다 약 {speedup:.1f}배 빠릅니다.")
```
